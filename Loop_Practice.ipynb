Nested for loop

```python
'''Nested for loop'''

List1 = ["Yellow", "Green", "Red"]
List2 = ["Pink", "Orange", "Yellow"]

for i in List1:
    for j in List2:
        if i==j:
            print("Matched: ",i,j)
```

```python
List01 = ["Red", "Yellow", "LightGreen"]
List02 = ["White", "Black", "Yellow"]
List03 = ["White", "Purple", "Yellow"]

for i in List01:
    for j in List02:
        for k in List03:
            if i==j==k:
                print("Matched: ",i,j,k)
```
