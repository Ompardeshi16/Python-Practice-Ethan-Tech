```python
# 1. Arithmetic Operators
 
a = 10
b = 3

print("Arithmetic Operators")
print(a + b)   # addition
print(a - b)   # subtraction
print(a * b)   # multiplication
print(a / b)   # division
print(a % b)   # modulus
print(a ** b)  # exponent
print(a // b)  # floor division
```

```python
# 2. Comparison Operators
 
print("\nComparison Operators")
print(a == b)  # equal to
print(a != b)  # not equal to
print(a > b)   # greater than
print(a < b)   # less than
print(a >= b)  # greater than or equal to
print(a <= b)  # less than or equal to
```

```python
# 3. Logical Operators
 
x = True
y = False

print("\nLogical Operators")
print(x and y)  # both must be True
print(x or y)   # at least one is True
print(not x)    # opposite of x
```

```python
# 4. Assignment Operators
 
print("\nAssignment Operators")
num = 5
num += 2    # num = num + 2
num -= 1    # num = num - 1
num *= 3    # num = num * 3
num /= 2    # num = num / 2
num %= 3    # num = num % 3
print(num)
```

```python
# 5. Bitwise Operators
 
p = 10  # 1010 in binary
q = 4   # 0100 in binary

print("\nBitwise Operators")
print(p & q)   # AND
print(p | q)   # OR
print(p ^ q)   # XOR
print(~p)      # NOT
print(p << 1)  # left shift
print(p >> 1)  # right shift
```

```python
# 6. Identity Operators
 
print("\nIdentity Operators")
m = [1, 2, 3]
n = m
o = [1, 2, 3]

print(m is n)   # same object
print(m is o)   # different object with same values
print(m is not o)
```

```python
# 7. Membership Operators
 
print("\nMembership Operators")
letters = ["a", "b", "c"]

print("a" in letters)
print("z" not in letters)
```
